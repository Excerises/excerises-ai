```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_rate = pd.read_csv("../datasets/gym_churn_rate.csv")
df_rate = pd.DataFrame(data={
    'age_range': df_rate['Age Range'],
    'total_count': df_rate['Total Count'],
    'total_churn': df_rate['Total Churn'],
    'churn_rate': df_rate['Churn Rate'],
})
df_rate
```

```python
for label in ['total_count', 'total_churn', 'churn_rate']:
    sns.barplot(df_rate.sort_values('age_range'), x="age_range", y=label)
    plt.show()
```
