```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, accuracy_score
import joblib

# Load dataset exercise_recommendation
df_exercise = pd.read_csv('../datasets/cleaned/exercise_recommendation.csv', index_col=0)
df_exercise.head()
```

```python
# Kolom fitur kriteria rekomendasi
feature_cols = ['body_part', 'equipment', 'target', 'secondary_muscles', 'difficulty', 'category']

# Menggunakan One-Hot Encoding untuk fitur kategorikal nominal
# Mencegah distorsi jarak akibat LabelEncoder numerik acak pada KNN
X = pd.get_dummies(df_exercise[feature_cols])

# LabelEncoder khusus untuk target output nama latihan ('name')
le_name = LabelEncoder()
y = le_name.fit_transform(df_exercise['name'])

print("Fitur setelah One-Hot Encoding (X shape):", X.shape)
print("Target label 'name' ter-encode (y shape):", y.shape)
print("Jumlah kelas nama latihan unik:", len(np.unique(y)))
X.head()
```

```python
# Training KNN dengan metric='cosine' pada SELURUH dataset (tanpa split)
# Metric Cosine sangat ideal untuk menghitung keserupaan vektor fitur biner One-Hot Encoding
X_train = X
y_train = y

knn_model = KNeighborsClassifier(n_neighbors=5, metric='cosine')
knn_model.fit(X_train, y_train)

print("Model KNN Rekomendasi Latihan (One-Hot Encoding + Cosine Metric) berhasil dilatih pada seluruh dataset!")
```

```python
# Prediksi Top-1 dan Top-5 pada Data Dummy Testing
distances, indices = knn_model.kneighbors(X_test_dummy, n_neighbors=5)
y_pred_top1 = knn_model.predict(X_test_dummy)

# 1. Hitung Exact Single Name Accuracy
acc_exact = accuracy_score(y_test_dummy, y_pred_top1)

# 2. Hitung Top-5 Recommendation Match Accuracy (Apakah ground truth name ada di 5 rekomendasi teratas?)
top5_matches = []
for i in range(len(dummy_test_data)):
    rec_names = df_exercise.iloc[indices[i]]['name'].values
    actual_name = dummy_test_data.iloc[i]['name']
    top5_matches.append(actual_name in rec_names)
acc_top5 = np.mean(top5_matches)

# 3. Hitung Criteria Match Accuracy (Apakah rekomendasi teratas 100% sesuai dengan kriteria input?)
criteria_matches = []
for i in range(len(dummy_test_data)):
    top1_rec_idx = indices[i][0]
    rec_feat = df_exercise.iloc[top1_rec_idx]
    act_feat = dummy_test_data.iloc[i]
    is_match = (
        rec_feat['body_part'] == act_feat['body_part'] and
        rec_feat['equipment'] == act_feat['equipment'] and
        rec_feat['target'] == act_feat['target'] and
        rec_feat['difficulty'] == act_feat['difficulty'] and
        rec_feat['category'] == act_feat['category']
    )
    criteria_matches.append(is_match)
acc_criteria = np.mean(criteria_matches)

print(f"=== HASIL EVALUASI MODEL REKOMENDASI LATIHAN ===")
print(f"1. Criteria Match Accuracy (Sesuai 100% Kriteria Input User) : {acc_criteria * 100:.2f}%")
print(f"2. Top-5 Recommendation Accuracy (Ground Truth dalam Top 5) : {acc_top5 * 100:.2f}%")

# Tabel Evaluasi Rekomendasi
results_df = pd.DataFrame({
    'Aktual (Ground Truth)': dummy_test_data['name'],
    'Rekomendasi Teratas (Top-1)': df_exercise.iloc[indices[:, 0]]['name'].values,
    'Sesuai Kriteria Input': criteria_matches,
    'Termasuk di Top-5 Rec': top5_matches
})

print("--- Tabel Hasil Evaluasi Data Dummy ---")
print(results_df.to_string())
```

```python
# Evaluasi Performa Model untuk berbagai Nilai K (1 hingga 15)
k_range = range(1, 16)
top_k_acc_list = []

for k in k_range:
    knn = KNeighborsClassifier(n_neighbors=k, metric='cosine')
    knn.fit(X_train, y_train)
    dists, inds = knn.kneighbors(X_test_dummy, n_neighbors=k)
    
    # Check Top-K match
    tk_match = [dummy_test_data.iloc[i]['name'] in df_exercise.iloc[inds[i]]['name'].values for i in range(len(dummy_test_data))]
    top_k_acc_list.append(np.mean(tk_match))

plt.figure(figsize=(9, 5))
plt.plot(k_range, [a * 100 for a in top_k_acc_list], marker='o', color='g', linestyle='-', label='Top-K Recommendation Accuracy')
plt.title('Akurasi Rekomendasi Latihan vs Nilai K (Metric: Cosine Similarity)')
plt.xlabel('Nilai K (n_neighbors)')
plt.ylabel('Akurasi (%)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Fungsi Rekomendasi Latihan Berdasarkan Input User
def recommend_exercise(body_part, equipment, target, secondary_muscles, difficulty, category, top_n=5):
    input_dict = {
        'body_part': [body_part],
        'equipment': [equipment],
        'target': [target],
        'secondary_muscles': [secondary_muscles],
        'difficulty': [difficulty],
        'category': [category]
    }
    input_df = pd.DataFrame(input_dict)
    input_encoded = pd.get_dummies(input_df).reindex(columns=X.columns, fill_value=0)
    
    distances, indices = knn_model.kneighbors(input_encoded, n_neighbors=top_n)
    recommendations = df_exercise.iloc[indices[0]][['name', 'body_part', 'equipment', 'target', 'difficulty', 'category']]
    return recommendations

# Contoh Penggunaan Rekomendasi
print("--- Contoh Hasil Rekomendasi Latihan ---")
sample_rec = recommend_exercise(
    body_part='waist',
    equipment='body weight',
    target='abs',
    secondary_muscles="['obliques']",
    difficulty='beginner',
    category='strength',
    top_n=5
)
sample_rec
```
