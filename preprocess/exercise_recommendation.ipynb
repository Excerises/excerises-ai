```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../datasets/exercise_recommendation.csv')

used_columns = ['bodyPart', 'equipment', 'id', 'name', 'target', 'secondaryMuscles', 'instructions', 'description', 'difficulty', 'category']
df = df[used_columns]

df.head()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
for label in ['bodyPart', 'equipment', 'target', 'difficulty', 'category']:
    sns.barplot(df[label])
    plt.show()
    print(df[label].value_counts())
```

```python
filtered_df = pd.DataFrame(data={
    'id': df['id'],
    'name': df['name'],
    'description': df['description'],
    'body_part': df['bodyPart'],
    'equipment': df['equipment'],
    'target': df['target'],
    'secondary_muscles': df['secondaryMuscles'],
    'instructions': df['instructions'],
    'difficulty': df['difficulty'].map(lambda x: str(x).lower()),
    'category': df['category'],
})
filtered_df.head()
```

```python
filtered_df.to_csv("../datasets/cleaned/exercise_recommendation.csv")
```
